```python
## Generado con IA --> Descargar de los datos utilizados

import os

# 1. Define the base path of the newly created volume
volumen_path = "/Volumes/ecommerce/source_data/raw/"

# 2. Dictionary containing the folder structure and download links
# IMPORTANT: Replace "archivo.csv" in the URLs with the exact file names from your GitHub repository
archivos_github = {
    "brands/brands.csv": "https://raw.githubusercontent.com/jilmecm/project_ecommerce/setup/0_data/raw/brands/archivo.csv",
    "category/category.csv": "https://raw.githubusercontent.com/jilmecm/project_ecommerce/setup/0_data/raw/category/archivo.csv",
    "customers/customers.csv": "https://raw.githubusercontent.com/jilmecm/project_ecommerce/setup/0_data/raw/customers/archivo.csv",
    "date/date.csv": "https://raw.githubusercontent.com/jilmecm/project_ecommerce/setup/0_data/raw/date/archivo.csv",
    "order_items/landing/landing.csv": "https://raw.githubusercontent.com/jilmecm/project_ecommerce/setup/0_data/raw/order_items/landing/archivo.csv",
    "products/products.csv": "https://raw.githubusercontent.com/jilmecm/project_ecommerce/refs/heads/main/0_data/raw/products/products.csv"
}

# 3. Iterate to create folders and download the files
for ruta_destino, url_origen in archivos_github.items():
    # Build the full path within Databricks
    ruta_completa = os.path.join(volumen_path, ruta_destino)
    
    # Create the directory (e.g., /Volumes/.../raw/brands) if it doesn't exist
    os.makedirs(os.path.dirname(ruta_completa), exist_ok=True)
    
    # Download the file using wget
    comando_descarga = f'wget -q -O "{ruta_completa}" "{url_origen}"'
    os.system(comando_descarga)
    
    print(f"Folder structure ready and file downloaded at: {ruta_completa}")
```
